# Tensor-Network Basics

This notebook demonstrates the basic tensor concepts used in matrix product
states and the density matrix renormalization group.

A rank-\(k\) tensor is an array with \(k\) indices.

| Object | Rank | Notation |
|---|---:|---|
| Scalar | 0 | \(c\) |
| Vector | 1 | \(v_i\) |
| Matrix | 2 | \(M_{ij}\) |
| Higher-rank tensor | \(k\geq3\) | \(A_{i_1\ldots i_k}\) |

In tensor-network diagrams:

- A tensor is represented by a node.
- Each tensor index is represented by a line.
- Connecting two lines means summing over the corresponding indices.
- Open lines represent the indices of the resulting tensor.

```python
import numpy as np
import dmrg
```

```python
scalar = 5

vector = np.array([1, 2, 3])

matrix = np.array([
    [1, 2],
    [3, 4],
])

higher_rank_tensor = np.zeros((2, 3, 4))


examples = {
    "scalar": scalar,
    "vector": vector,
    "matrix": matrix,
    "higher-rank tensor": higher_rank_tensor,
}


for name, example in examples.items():
    information = dmrg.tensors.describe_tensor(example)

    print(f"{name}:")
    print(f"  type = {information['type']}")
    print(f"  rank = {information['rank']}")
    print(f"  notation = {information['notation']}")
    print(f"  dimensions = {information['dimensions']}")
    print(f"  number of indices = {information['number_of_indices']}")
    print(f"  number of elements = {information['number_of_elements']}")
    print()
```
